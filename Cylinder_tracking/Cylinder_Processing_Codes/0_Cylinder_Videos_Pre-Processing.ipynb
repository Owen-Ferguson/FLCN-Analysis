## This code pre-processes your videos for faster analysis within DeepLabCut

Change the path here to your videos and Run All to convert your videos for easier analysis

Make sure that your kernel is set to DEEPLABCUT and not base

```python
path = r""
```

No need to change any paths here; the new videos will be saved in your input folder

```python
import os
import subprocess

# Output folder for converted videos (optional)
output_path = os.path.join(path, 'DLC_Converted_Videos')
os.makedirs(output_path, exist_ok=True)

# Loop through all video files in the folder
for file in os.listdir(path):
    if file.lower().endswith(('.mp4', '.mov', '.avi', '.mkv')):  # adjust as needed
        input_file = os.path.join(path, file)
        output_file = os.path.join(output_path, f"{os.path.splitext(file)[0]}_480p.mp4")

        # FFmpeg command
        cmd = [
            'ffmpeg',
            '-i', input_file,
            '-vf', 'scale=-2:480,fps=10',
            '-an',
            output_file
        ]

        print(f"Converting: {file}")
        subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        



print("✅ All videos converted.")
```
